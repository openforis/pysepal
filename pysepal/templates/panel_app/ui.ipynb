```python
from pysepal import sepalwidgets as sw

from component.message import messages
from component.tile.widget_tile import WidgetTile
```

```python
# Create an appBar
app_bar = sw.AppBar(messages.msg("app.title"))
```

```python
# Create a footer
app_footer = sw.Footer(messages.msg("app.footer", year=2020))
```

```python
# load the partial files
%run "about_ui.ipynb"
widget_tile = WidgetTile()
# Gather all the partial tiles that you created previously
app_content = [about_tile, disclaimer_tile, widget_tile]
```

```python
# create a drawer for each group of tile
# fmt: off
items = [
    sw.DrawerItem(messages.msg("app.drawer_item.about"), "fa-solid fa-question-circle", card="about_tile"),
    sw.DrawerItem(messages.msg("app.drawer_item.widgets"), "fa-solid fa-wrench", card="widget_tile"),
]
# fmt: on

# Add the links to the code, wiki and issue tracker of your
code_link = "https://github.com/openforis/pysepal/"
wiki_link = "https://github.com/openforis/pysepal/blob/main/doc/en.rst"
issue_link = "https://github.com/openforis/pysepal/issues/new"

# Create the side drawer with all its components
app_drawer = sw.NavDrawer(items=items, code=code_link, wiki=wiki_link, issue=issue_link)
```

```python
# build the Html final app by gathering everything
app = sw.App(
    tiles=app_content,
    appBar=app_bar,
    footer=app_footer,
    navDrawer=app_drawer,
)
```

```python
# display the app
app.show_tile("about_tile").add_banner(messages.msg("app.banner"))
```
